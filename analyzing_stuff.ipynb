```python
def f(en, po):
    return 1/(np.exp(en+po)+1)
    
```

```python
def f_e(en):
    return 1/(np.exp(en/32)+1)
```

```python
def F_LR(p1, e2, e3, p4):
    gl = -0.27663
    return f_e(e3)*(1-f_e(e2))*((gl+1)*f(p4,0.01)*(1-f(p1,0.01)) + gl*f(p4,-0.01)*(1-f(p1,-0.01))) #- f_e(e2)*(1-f_e(e3))*((gl+1)*(1-f(p4,0.01))*f(p1,0.01)+gl*(1-f(p4,-0.01))*f(p1,-0.01))
```
