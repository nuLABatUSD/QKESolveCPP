```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
vz = np.array(pd.read_csv("v_z_outputs.csv",header=None))
vmag = np.array(pd.read_csv("v_mags_outputs.csv",header=None))
time = np.array(pd.read_csv("v_m_times.csv",header=None))
```

```python
vzs = []
vmags = []
times = []
for i in range(len(vz[0])-1):
    vzs.append(vz[0][i])
    vmags.append(vmag[0][i])
    times.append(time[0][i])
```

```python
plt.figure()
plt.plot(times[:20], vzs[:20], "-o")
plt.title("v_density z component vs time")
plt.show()
```

```python
plt.figure()
plt.plot(times, vmags, label="v_density")
plt.plot(times,np.zeros(len(times)), label="v_vacuum")
plt.title("v magnitude vs time")
plt.legend()
plt.show()
```
